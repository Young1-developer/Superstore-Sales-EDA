IMPORTING THE TRINITY OF DATA ANALYSIS

```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
```

READING AND SEEYING DATASET

```python
dataset = pd.read_csv('train.csv')
df = dataset.copy()
df.shape
```

SEEYING THE COLUMNS

```python
df.columns
```

STANDARDISING THE HEADING 

```python
df.columns = (
    df.columns.str.strip()
    .str.replace(' ', '_')
    .str.lower()
)
```

SEEYING THE CHANGE

```python
df.head()
```

CHECKING DATA TYPES

```python
df.dtypes
```

REMOVING FISRT COLUMN

```python
df.drop(df.columns[0], axis=1, inplace=True)
```

CHECKING FOR DUPLICATE

```python
df.duplicated().sum()
```

DELETING DUPLICATE

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

CHECKING FOR NAN

```python
df.isna().sum()
```

WHAT ARE THR TOP TEN PRODUCT BY CATEGORY THAT HAVE HIGH SALES?

```python
df[df['category'] == 'Furniture'].sort_values(by='sales', ascending=False).head(10)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['ship_date'] = pd.to_datetime(df['ship_date'], errors='coerce')
```

```python
df['order_date'].min(), df['order_date'].max() 
```

filter top 10 most recent orders

```python
df.loc[:10,['customer_name','category','city','country','order_date','ship_date','sales']].sort_values(by='order_date', ascending=False)
```

```python
sales_by_stateUSA = df[df['country'] == 'United States'].groupby('state')['sales'].sum().sort_values(ascending=False).head(10)
sales_by_stateUSA.plot(kind='bar', x='state', y='sales', title='Top 10 States by Sales in the United States', xlabel='State', ylabel='Sales', color='lightgreen')    
```

```python
df['category'].value_counts().plot(kind='bar',title='Sales by Category', xlabel='Category', ylabel='Count', color='lightblue')
```

```python
df.drop(columns='total_sales_by_region',inplace=True)
```

```python
df.head()
```

```python
total_sales_by_region = df.groupby('region')['sales'].sum()
total_sales_by_region
```

```python
total_sales_by_region.plot(kind='bar', title='Total Sales by Region', xlabel='Region', ylabel='Total Sales', color='lightcoral')
```

```python
total_sales_by_state = df.groupby('state')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
total_sales_by_state.plot(kind='bar', title='Total Sales by State', xlabel='State', ylabel='Total Sales', color='lightcoral')
```

```python
top_orders = df['order_id'].value_counts().sort_values(ascending=False).head(10)
top_orders.plot(kind='bar',title='To 10 Orders By Customers',xlabel='customersID',ylabel='Number of Orders',color='lightblue')
```

```python
top_orders.reset_index()
```

AVERAGE SALES PER SEGMENT IN UNITED STATES

```python
average_sales_segmentUSA = df[df['country'] == 'United States'].groupby('segment')['sales'].mean().sort_values(ascending=False)
average_sales_segmentUSA.plot(kind='bar',title='Average Sales per Segment in USA',xlabel='Customer Sgement',ylabel='sales($)',color='lightgreen')
```
